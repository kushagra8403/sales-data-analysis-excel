# Customer Churn & Retention Analysis

**Business question:** Where is recurring revenue most exposed to churn, and which customer segments should be investigated first?

> Portfolio data is synthetic and used for demonstration.

```python
import pandas as pd
from pathlib import Path

path = Path('../02-customer-churn-analysis/data/customers.csv')
df = pd.read_csv(path)
df.head()
```

```python
kpis = {
    'Customers': len(df),
    'Churn rate': df['churned'].mean(),
    'Active MRR': df.loc[df.churned.eq(0), 'monthly_charge'].sum(),
    'Monthly revenue at risk': df.loc[df.churned.eq(1), 'monthly_charge'].sum()
}
pd.Series(kpis)
```

```python
import matplotlib.pyplot as plt

plan = df.groupby('plan')['churned'].mean().sort_values(ascending=False)
plan.plot(kind='bar', title='Churn rate by plan', ylabel='Churn rate', xlabel='Plan')
plt.tight_layout()
plt.show()
```

## Business interpretation

Use the KPI output and segment chart to identify plans and tenure groups with elevated churn. The next operational step would be to validate those patterns against real customer behaviour, service contacts and retention activity before making commercial changes.
